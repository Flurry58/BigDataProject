```python
import requests
import json
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from dotenv import load_dotenv
import datetime
import json 
```

```python
datapath = "2017/"
start_date = "2017-07-05"
end_date = "2017-10-10"

file_names = []
while start_date != end_date:
    
    file_names.append(start_date + ".txt")
    str_date = start_date.split("-")
    year = int(str_date[0])
    month = int(str_date[1])
    day = int(str_date[2])

    date_start_date = datetime.date(year, month, day)
    date_start_date += datetime.timedelta(days=1)

    start_date = date_start_date.strftime("%Y-%m-%d")

file_names.append(end_date + ".txt")
```

```python
df_list = []
for name in file_names:
    with open(datapath + name) as f:
    	text = f.readlines()

    lst = [json.loads(line) for line in text]
    
    df_list.append(pd.json_normalize(lst))
```

```python
weather_data = pd.concat(df_list)
```

```python
# hopefully this works
weather_data["properties.created"] = pd.to_datetime(weather_data["properties.created"], format='ISO8601')
weather_data["properties.from"] = pd.to_datetime(weather_data["properties.from"], format='ISO8601', utc=True)
weather_data["properties.to"] = pd.to_datetime(weather_data["properties.to"], format='ISO8601', utc=True)
weather_data["properties.calculatedAt"] = pd.to_datetime(weather_data["properties.calculatedAt"], format='ISO8601')
```
